```python
# class: bot viết csv từ tọa độ các điểm bàn tay và class của bàn tay
class CSVWriterBot:
    def __init__(self, PATH):
        self.csv_file = open(PATH, "a")
        self.writer = csv.writer(self.csv_file, delimiter=',', quotechar='|',quoting=csv.QUOTE_MINIMAL)
    
    def add_row(self, hand_gestures, label):
        # viết thêm một hàng mới bằng writer
        self.writer.writerow([label, *np.array(hand_gestures).flatten().tolist()])   # unlist
    
    def close(self):
        self.csv_file.close()
```
